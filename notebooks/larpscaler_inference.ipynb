# LarpScaler inference from Hugging Face

This notebook installs the standalone GitHub package, downloads the inference-only checkpoint from Hugging Face, and upscales one image.

```python
# For Colab/Jupyter after the GitHub repository is published:
%pip install -q git+https://github.com/Vovanm88/LARP-Scaler.git
```

```python
from pathlib import Path
import torch
from IPython.display import display
from PIL import Image
from larpscaler import LarpScaler

MODEL_ID = "VladimirM388/larpscaler-v2-bf16"
INPUT_PATH = Path("input.png")
OUTPUT_PATH = Path("output.png")
print("CUDA available:", torch.cuda.is_available())
```

Upload or copy an image to `input.png`. For a private Hugging Face repository, run `huggingface_hub.login()` once before loading.

```python
source = Image.open(INPUT_PATH).convert("RGB")
display(source)
print("Input size:", source.size)
```

```python
# The first call downloads transformer/, scheduler/, vae/, and conditioning.
# Later calls reuse the normal Hugging Face cache.
upscaler = LarpScaler.from_pretrained(MODEL_ID)
```

```python
result = upscaler.upscale(
    source,
    scale=4,
    steps=1,
    noise_level=0.7,
    guidance_scale=4.5,
    seed=1234,
)
display(result)
print("Output size:", result.size)
```

```python
result.save(OUTPUT_PATH)
print("Saved to", OUTPUT_PATH.resolve())
```

## Optional fast preset

For lower latency, use `steps=1`, `noise_level=0.35`, and `guidance_scale=1.0`.
